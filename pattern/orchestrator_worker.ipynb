```python
from langchain_core.tracers import ConsoleCallbackHandler
import operator
from typing import List, TypedDict, Annotated

from IPython.core.display import Image, Markdown
from IPython.core.display_functions import display
from langchain_core.messages import SystemMessage, HumanMessage!
from langgraph.constants import START, END
from langgraph.graph import StateGraph
from langgraph.types import Send
from pydantic import BaseModel, Field

from utils.std_model import base_model

llm = base_model()


# Schema for structured output to use in planning
class Section(BaseModel):
    name: str = Field(
        description="段落的标题",
    )
    description: str = Field(
        description="简单描述段落的主要内容和概念。",
    )


class Sections(BaseModel):
    sections: List[Section] = Field(
        description="报告的段落集合",
    )


class State(TypedDict):
    topic: str
    sections: List[Section]
    completed_sections: Annotated[
        list, operator.add
    ]
    final_report: str


# 这里 WorkerState 只是 State 的投影，没有实质作用，只是说明在 work_call 只需要这两个
class WorkerState(TypedDict):
    section: Section
    completed_sections: Annotated[list, operator.add]


def planner_call(state: State):
    planner = llm.with_structured_output(Sections)

    resp = planner.invoke(
        [
            SystemMessage(content="请根据要求整理生成完成报告的计划"),
            HumanMessage(content=f"这个报告的主题是[{state['topic']}]"),
        ]
    )

    return {
        "sections": resp.sections,
    }


def assign_workers(state: State):
    return [Send(worker_call.__name__, {"section": s}) for s in state["sections"]]


def worker_call(state: WorkerState):
    resp = llm.invoke(
        [
            SystemMessage(content="按照提供的名称和描述来写报告的段落。段落前面不要加开场白。使用Markdown格式排版。"),
            HumanMessage(content=f"段落的名称[{state["section"].name}],描述[{state["section"].description}]"),
        ]
    )

    return {
        "completed_sections": [resp.content],
    }


def sync_worker(state: State):
    completed_sections = state["completed_sections"]
    completed_report_sections = "\n\n---\n\n".join(completed_sections)
    return {"final_report": completed_report_sections}


graph_builder = StateGraph(State)

(graph_builder
 .add_node(planner_call.__name__, planner_call)
 .add_node(worker_call.__name__, worker_call)
 .add_node(sync_worker.__name__, sync_worker)
 )

(graph_builder
 .add_edge(START, planner_call.__name__)
 .add_conditional_edges(planner_call.__name__, assign_workers, [worker_call.__name__])
 .add_edge(worker_call.__name__, sync_worker.__name__)
 .add_edge(sync_worker.__name__, END)
 )

graph = graph_builder.compile()

# Show the workflow
display(Image(graph.get_graph().draw_mermaid_png()))

graph_resp = graph.invoke({
    "topic": "请生成一个有关东莞餐饮类别分布的报告"
}, config={"callbacks": [ConsoleCallbackHandler()]})

Markdown(graph_resp["final_report"])
```
